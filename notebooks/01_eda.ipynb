# VN2 Inventory Data — Exploratory Analysis

Weekly sales panels for 599 store-product pairs (2021-04-12 → 2024-06-03).

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from src.forecasting.preprocessing import load_panel_from_config
from src.utils.config import load_config

sns.set_theme(style="whitegrid")
config = load_config(ROOT / "configs" / "config.yaml")
panel = load_panel_from_config(config)
panel.head()
```

```python
print(f"Rows: {len(panel):,}")
print(f"Store-products: {panel.groupby(['store','product']).ngroups}")
print(f"Date range: {panel['date'].min()} → {panel['date'].max()}")
print(f"OOS rate: {(~panel['in_stock']).mean():.2%}")
print(f"Zero sales (in-stock): {((panel['sales']==0) & panel['in_stock']).mean():.2%}")
```

```python
weekly = panel.groupby('date')['sales'].sum(min_count=1).reset_index()
fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(weekly['date'], weekly['sales'])
ax.set_title('Total weekly sales (all SKUs)')
ax.set_xlabel('Week')
plt.tight_layout()
plt.show()
```

```python
sku = panel[(panel['store']==2) & (panel['product']==126)].dropna(subset=['sales'])
fig, ax = plt.subplots(figsize=(12, 3))
ax.plot(sku['date'], sku['sales'], marker='o', markersize=2)
ax.set_title('Example SKU: store=0, product=126')
plt.tight_layout()
plt.show()
```
